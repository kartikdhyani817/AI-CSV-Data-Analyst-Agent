```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]


df = pd.read_csv(filename, encoding='latin1')
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()
df.describe()
```

```python
print("Missing Values")

print(df.isnull().sum())
```

```python
print("Column Names")

print(df.columns.tolist())
```

```python
summary = f"""
Dataset Summary

Rows: {df.shape[0]}

Columns: {df.shape[1]}

Columns Available:

{df.columns.tolist()}
"""

print(summary)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)

df = df.drop_duplicates()

print("Duplicates Removed")
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

print(missing_percent)
```

```python
print("DATA QUALITY REPORT")

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing Values")

print(df.isnull().sum())

print("\nDuplicate Rows")

print(df.duplicated().sum())
```

```python
numerical_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns

print(numerical_columns)
```

```python
df[numerical_columns].describe()
```

```python
correlation = df[numerical_columns].corr()

correlation
```

```python
corr_pairs = correlation.unstack()

corr_pairs = corr_pairs.sort_values(
    ascending=False
)

print(corr_pairs.head(20))
```

```python
categorical_columns = df.select_dtypes(
    include=['object']
).columns

for col in categorical_columns:

    print("\n", col)

    print(
        df[col].value_counts().head(5)
    )
```

```python
insights = []

insights.append(
    f"Dataset contains {df.shape[0]} rows."
)

insights.append(
    f"Dataset contains {df.shape[1]} columns."
)

insights.append(
    f"Numerical Columns: {len(numerical_columns)}"
)

insights.append(
    f"Categorical Columns: {len(categorical_columns)}"
)

print("KEY INSIGHTS")

for i in insights:
    print("-", i)
```

```python
report = "\n".join(insights)

with open(
    "dataset_insights.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(report)

print("Report Saved")
```
